```python
# Ejercicio 1: Calculando el salario semanal
# En este ejercicio se solicita calcular el salario total de un vendedor. El vendedor recibe un salario base de $500 más una comisión del 10% sobre el total de ventas realizadas.

# Solicitar las ventas totales al usuario
ventas_totales = float(input("Ingrese el total de ventas del vendedor: "))

# Definir salario base
salario_base = 500

# Calcular comisión
comision = ventas_totales * 0.10

# Calcular salario total
salario_total = salario_base + comision

# Mostrar resultado
print(f"El salario total del vendedor es: ${salario_total}")
```

```python
# Ejercicio 2: Bienvenido a la frutería virtual
# En este ejercicio se utiliza un diccionario para almacenar frutas y sus precios.
# El programa solicita al usuario ingresar una fruta y luego muestra su precio si existe en el catálogo.

# Crear diccionario de frutas
catalogo_frutas = {
    "manzana": 1200,
    "banana": 900,
    "naranja": 1000,
    "pera": 1500
}

# Solicitar fruta al usuario
fruta = input("Ingrese el nombre de una fruta: ")

# Verificar si la fruta existe
if fruta in catalogo_frutas:
    print(f"El precio de la {fruta} es ${catalogo_frutas[fruta]}")
else:
    print("La fruta no se encuentra en nuestra tienda.")
```

```python
# Ejercicio 3: Bienvenido al casino
# En este ejercicio se utiliza la librería random para simular el lanzamiento de dos dados.
# El programa genera números aleatorios entre 1 y 6 y luego muestra la suma total obtenida.

# Importar librería random
import random

# Generar valores aleatorios para los dados
dado1 = random.randint(1, 6)
dado2 = random.randint(1, 6)

# Calcular suma
suma = dado1 + dado2

# Mostrar resultados
print(f"El valor del primer dado es: {dado1}")
print(f"El valor del segundo dado es: {dado2}")
print(f"La suma de ambos dados es: {suma}")
```

```python
# Ejercicio 4: Calculadora de chocolate
# En este ejercicio el programa calcula la cantidad máxima de porciones de chocolate que una persona puede consumir sin superar su límite diario de calorías.

# Solicitar datos al usuario
limite_calorias = int(input("Ingrese su límite diario de calorías: "))
calorias_chocolate = int(input("Ingrese las calorías de una porción de chocolate: "))

# Calcular cantidad máxima de porciones
cantidad_maxima = limite_calorias // calorias_chocolate

# Mostrar resultado
print(f"Puedes comer hasta {cantidad_maxima} porciones de chocolate sin exceder tu límite diario.")
```
