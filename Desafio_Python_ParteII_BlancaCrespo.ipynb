## Ejercicio 1: Suma de números pares
En este ejercicio se solicitarán dos números enteros al usuario.
Luego, el programa recorrerá todos los números que se encuentran entre ambos valores utilizando la función `range()`.
Finalmente, se sumarán solamente los números pares y se mostrará el resultado.

print("Ejercicio 1: Suma de números pares")

numero_1 = int(input("Ingrese el primer número entero: "))
numero_2 = int(input("Ingrese el segundo número entero: "))

inicio = min(numero_1, numero_2)
fin = max(numero_1, numero_2)

suma_pares = 0

for numero in range(inicio, fin + 1):
    if numero % 2 == 0:
        suma_pares += numero

print("La suma de los números pares entre", inicio, "y", fin, "es:", suma_pares)

## Ejercicio 2: Calculadora de áreas
En este ejercicio se solicitará al usuario el tipo de figura geométrica de la que desea calcular el área.
El programa permitirá calcular el área de un cuadrado, un triángulo o un círculo.
Para resolverlo se utilizarán variables, entrada de datos con `input()`, condiciones `if`, `elif`, `else` y la librería `math` para obtener el valor de pi en el cálculo del área del círculo.

```python
# Ejercicio 2: Calculadora de áreas

import math

print("Ejercicio 2: Calculadora de áreas")
print("Figuras disponibles: cuadrado, triangulo, circulo")

figura = input("Ingrese la figura geométrica: ").lower()

if figura == "cuadrado":
    lado = float(input("Ingrese la medida del lado del cuadrado: "))
    area = lado ** 2
    print("El área del cuadrado es:", area)

elif figura == "triangulo" or figura == "triángulo":
    base = float(input("Ingrese la base del triángulo: "))
    altura = float(input("Ingrese la altura del triángulo: "))
    area = (base * altura) / 2
    print("El área del triángulo es:", area)

elif figura == "circulo" or figura == "círculo":
    radio = float(input("Ingrese el radio del círculo: "))
    area = math.pi * radio ** 2
    print("El área del círculo es:", area)

else:
    print("La figura ingresada no es válida.")
```

## Ejercicio 3: Adivina el número
En este ejercicio el programa generará un número aleatorio entre 1 y 100.
Luego, el usuario deberá intentar adivinar el número. El programa entregará pistas indicando si el número ingresado es mayor o menor que el número secreto.
Para resolverlo se utilizará la librería `random`, una variable para guardar el número secreto, condiciones `if`, `elif`, `else` y un ciclo `while` para mantener el programa activo hasta que el usuario adivine correctamente.

```python
# Ejercicio 3: Adivina el número

import random

print("Ejercicio 3: Adivina el número")

numero_secreto = random.randint(1, 100)
intentos = 0

while True:
    numero_usuario = int(input("Ingrese un número entre 1 y 100: "))
    intentos += 1

    if numero_usuario < numero_secreto:
        print("El número secreto es mayor.")

    elif numero_usuario > numero_secreto:
        print("El número secreto es menor.")

    else:
        print("¡Correcto! Adivinaste el número.")
        print("Cantidad de intentos:", intentos)
        break
```

## Ejercicio 4: Carrera de pollos
En este ejercicio se simulará segundo a segundo la carrera de un pollo que debe recorrer una distancia de 100 metros.
El programa solicitará al usuario la velocidad del pollo en metros por segundo. Luego, usando un ciclo `while`, irá aumentando la distancia recorrida en cada segundo.
En cada impresión se mostrará el emoji del pollo desplazándose hacia la derecha, hasta que alcance o supere los 100 metros.

# Ejercicio 4: Carrera de pollos

print("Ejercicio 4: Carrera de pollos")

velocidad = float(input("Ingrese la velocidad del pollo en metros por segundo: "))

distancia = 0
segundo = 0

while distancia < 100:
    segundo += 1
    distancia += velocidad

    espacios = int(distancia / 2)

    print("Segundo:", segundo, "| Distancia:", distancia, "metros")
    print(" " * espacios + "🐥")

print("¡El pollo llegó a la meta!")
print("Tiempo total:", segundo, "segundos")
